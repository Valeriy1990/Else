## Что такое chatGPT?

В данном уроке мы начнем знакомиться с chatGPT, который произвел революцию в области искусственного интеллекта и является ярким представителем генеративно-состязательных сетей. ChatGPT, разработанный OpenAI, основан на передовых технологиях, позволяющих создавать тексты и вести беседы, приближенные к человеческому стилю. Этот мощный инструмент привлек внимание множества областей, от разработки программного обеспечения до обучения и общения с искусственным интеллектом.

* Во-первых, chatGPT основан на технологии генеративно-состязательных сетей (GANs), что представляет собой форму машинного обучения, где две модели, генератор и дискриминатор, соперничают между собой, обучаясь и улучшаясь в процессе. * Во-вторых, использование трансформерной архитектуры, внедренной в chatGPT, позволяет обрабатывать и анализировать длинные последовательности данных с впечатляющей точностью. Это существенно улучшило способность модели к восприятию контекста и последовательности слов, что сделало ее более эффективной в обработке естественного языка.

* Также стоит отметить применение механизма внимания, который позволяет модели фокусироваться на ключевых элементах входных данных, учитывая их важность в формировании ответа. Этот механизм улучшает способность chatGPT к точной генерации ответов на запросы и ведению бесед в естественной форме.

````python
# Импортируем библиотеки:
# Для вызова OpenAI API
from openai import OpenAI
# Для работы с системными командами
import os
import getpass

# Выбираем модель GPT
GPT_MODEL = "gpt-3.5-turbo"
````

````python
# Помещаем OpenAI ключ в переменную окружения с именем 'OPENAI_API_KEY'
# Привыкайте использовать переменные окружения в своем коде. Это вам позволит легко переносить код на удаленный сервер со своим окружением.

os.environ["OPENAI_API_KEY"] = getpass.getpass("Введите OpenAI API Key:")
````

````python

# Запрос к chatGPT
query = 'Когда проходил чемпионат?'
# Создаём объект OpenAI
openai = OpenAI(
  api_key = os.environ.get("OPENAI_API_KEY"),
)
# Отправляем запрос к OpenAI API
response = openai.chat.completions.create(
    messages=[
        {'role': 'system', 'content': 'Вы отвечаете на вопросы о Чемпионате мира по крикету 2023'},
        {'role': 'user', 'content': query},
    ],
    model=GPT_MODEL,
    temperature=0,
)

# Выводим результат
print(response.choices[0].message.content)
````

В этом Блокноте мы рассмотрим двухэтапный метод Search-Ask и применим его к нашей задачи. Метод позволяет GPT отвечать на вопросы, используя библиотеку справочного текста:

* На первом этапе (Search) мы ищем в библиотеке текстов соответствующие текстовые разделы.
* На втором этапе (Ask) мы вставляем извлеченные текстовые фрагменты в сообщение для GPT и задаем ему вопрос.

Тонкая настройка лучше подходит для обучения специализированным задачам или стилям и менее надежна для запоминания фактов. Веса модели подобны долговременной памяти. Когда вы точно настраиваете модель, это похоже на подготовку к экзамену через неделю. Когда наступает время экзамена, модель может "забыть детали".

Напротив, ввод сообщений подобен кратковременной памяти. Когда вы вставляете знания в сообщение, то это похоже на сдачу экзамена со шпаргалкой в руке, даже если модель никогда не встречала этих данных, все равно шансов сдать экзамен значительно больше.

В методе `Search-Ask` поиск по тексту может осуществляться различными способами, например:

* Поиск на основе лексики (Lexical-based search)
* Поиск на основе графов (Graph-based search)
* Поиск на основе эмбендингов (Embedding-based search)

В этом примере блокнота используется поиск на основе [эмбендингов](https://platform.openai.com/docs/guides/embeddings/what-are-embeddings). Эмбендинги просты в реализации и особенно хорошо работают с вопросами, поскольку вопросы часто лексически не совпадают с их ответами.

Поисковые системы могут сочетать несколько методов поиска, а производительность поиска вопросов и ответов может быть улучшена с помощью таких методов, как [HyDE](https://arxiv.org/abs/2212.10496), в которых вопросы сначала преобразуются в гипотетические ответы, а затем внедряются.

Точные затраты будут зависеть от специфики системы и выбранной модели использования.

Токен - это минимальная единица, на которую разбивается входная текстовая последовательность. Мы с вами скоро дойдем до темы токинизации текстовых последовательностей и вы все подробно изучите. Поэтому часто прибегают к хитрости: переводят запрос на английский язык, а после ответ на русский. Так и дешевле, и выше качество ответов.

Чтобы определить число токенов существует удобная библиотека `tiktoken`.

С помощью нее напишем 2 вспомагательные функции:
* num_tokens - для определения числа токенов в строке с учетом заданной модели
* clip_tokens - для обрезки строки до заданного числа токенов и с учетом модели

````python
# устанавливаем библиотеку
!pip install tiktoken
````

````python
import tiktoken

# Функция возвращает число токенов в строке для заданной модели
def num_tokens(text: str, model: str = GPT_MODEL) -> int: # Новая для вас конструкция `-> int` указывает тип возвращаемого значения функции
    # получаем токинизатор для заданной модели
    encoding = tiktoken.encoding_for_model(model)
    # переводим текст в набор токенов (кодируем с помощью токенизатора)
    return len(encoding.encode(text))

# Функция обрезает строку до заданного числа токенов, учитывая модель
def clip_tokens(text: str, length: int = 4096, model: str = GPT_MODEL) -> str: # Новая для вас конструкция `-> str` указывает тип возвращаемого значения функции
    # получаем токинизатор для заданной модели
    encoding = tiktoken.encoding_for_model(model)

    # переводим текст в набор токенов (кодируем с помощью токенизатора)
    tokens=encoding.encode(text=text)

    # с помощью генератора цикла декодируем обратно нужное число токенов (4096 предел для нашей модели)
    # так как мы игнорируем при декодировании все символы, которые невозможно преобразовать (в статье попадаются фразы на языке Хинди), то токенов будет меньше 4096
    token_value = [
            encoding.decode_single_token_bytes(token).decode("utf-8", errors="ignore")
            for token in tokens[0:length]
        ]

    # объединяем элементы списка в строку
    return "".join(token_value)


````

Мы еще с вами будем разбирать, что такое токенизатор и как происходит токенизация текста. В упрощенном виде эту процедуру можно представить в следующем виде:
1. Определяется частотность слов в обучающей выборки для модели.
2. Слова ранжируются по частоте использования. Самые редкие слова в конце.
3. Каждому слову присваивается число. Чем реже слово встречается, тем больше число.
4. Любой текст, который подается модели преобразуется в числовую последовательность, происходит токенизация текста.  

На самом деле, описанный выше метод - это только одна из возможностей токенизаци, ведь можно токенизировать и по буквам, и по морфемам. Основная идея токенизации - преобразовать текст в последовательность цифр наиболее оптимальным образом. Но об этом чуть позже.

Выше мы определили 2 вспомогательные функции: `num_tokens` - для определения числа токенов в тексте и `clip_tokens` -  для обрезки текста до максимального числа токенов. Применим их к нашей статье:

````python
print(f'Число токенов в статье до обрезки: {num_tokens(page.content)}')
print(f'Число токенов в статье после обрезки: {num_tokens(clip_tokens(page.content))}')
print(f'Число символов в статье после обрезки: {len(clip_tokens(page.content))}')
````

Теперь мы можем передать нашу статью из Википедии вместе с основным запросом в OpenAI API:

````python
# Создаём объект OpenAI
openai = OpenAI(
  api_key = os.environ.get("OPENAI_API_KEY"),
)

query_gpt = f"""Используйте приведенную ниже статью для ответа на вопрос. Отвечай кратко. Если ответ не найден, напишите "Я не знаю."
Вопрос: Когда проводился чемпионат и кто в нем победил?
Статья:
\"\"\"
{query}
\"\"\"

"""


response = openai.chat.completions.create(
    messages=[
        {'role': 'system', 'content': 'Ты отвечаешь на вопросы о Чемпионате мира по крикету 2023'},
        {'role': 'user', 'content': clip_tokens(query_gpt)},
    ],
    model=GPT_MODEL,
    temperature=0,
)

print(response.choices[0].message.content)
````

#### 2. Search (Поиск)

Далее, определим функцию поиска, которая:

* Принимает пользовательский запрос и DataFrame со столбцами `text` и `embedding`;
* Токенизирует пользовательский запрос с помощью OpenAI API (для этого используется токенизатор `text-embedding-ada-002`);
* Использует косинусное расстояние для определения схожести между токенизированым пользовательским запросом и эмбендингами в DataFrame для ранжирования текстов по схожести или релевантности;
* Возвращает два списка:
  * N лучших текстов, ранжированных по релевантности;
  * Их соответствующие оценки релевантности.

````python
from scipy import spatial  # вычисляет сходство векторов
EMBEDDING_MODEL = "text-embedding-ada-002"

# Функция поиска
def strings_ranked_by_relatedness(
    query: str, # пользовательский запрос
    df: pd.DataFrame, # DataFrame со столбцами text и embedding (база знаний)
    relatedness_fn=lambda x, y: 1 - spatial.distance.cosine(x, y), # функция схожести, косинусное расстояние
    top_n: int = 100 # выбор лучших n-результатов
) -> tuple[list[str], list[float]]: # Функция возвращает кортеж двух списков, первый содержит строки, второй - числа с плавающей запятой
    """Возвращает строки и схожести, отсортированные от большего к меньшему"""

    # Отправляем в OpenAI API пользовательский запрос для токенизации
    query_embedding_response = openai.embeddings.create(
        model=EMBEDDING_MODEL,
        input=query,
    )

    # Получен токенизированный пользовательский запрос
    query_embedding = query_embedding_response.data[0].embedding

    # Сравниваем пользовательский запрос с каждой токенизированной строкой DataFrame
    strings_and_relatednesses = [
        (row["text"], relatedness_fn(query_embedding, row["embedding"]))
        for i, row in df.iterrows()
    ]

    # Сортируем по убыванию схожести полученный список
    strings_and_relatednesses.sort(key=lambda x: x[1], reverse=True)

    # Преобразовываем наш список в кортеж из списков
    strings, relatednesses = zip(*strings_and_relatednesses)

    # Возвращаем n лучших результатов
    return strings[:top_n], relatednesses[:top_n]
````

Разберем, что такое косинусное расстояние.

**Косинусное расстояние** — это угол между векторами пространства внутреннего произведения. Он определяет, указывают ли векторы в одном направлении. Косинусное расстояние можно использовать тогда, когда величина векторов не имеет значения.



image.png

Предположим, что $A$ и $B$ — две точки на двумерной плоскости.

$A = (X1, Y1)$

$B = (X2, Y2)$

$cosθ = (a · b) / (|a| * |b|)$

$(a · b)$ — скалярное произведение векторов $a$ и $b$.

$|а|$ — длина вектора $а$.

$|b|$ - длина вектора $b$.

$(|a| * |b|)$ — векторное произведение векторов $a$ и $b$.

Формула косинусного расстояния:

$d = 1 — cos θ$

$d = 1 — (a · b) / (|a| * |b|)$

>**Пример**. Предположим, что $(5,3)$ и $(2,4)$ — две точки на двумерной плоскости.
  $
  (a \cdotp b) = (5 * 2) + (3 * 4) = 10 + 12 = 22
  \\
  |a| = \sqrt{5^2 + 3^2} = 5.83
  \\
  |b| = \sqrt{2^2 + 4^2} = 4.47
  \\
  \\
  \text{ Используя формулу косинусного расстояния: }
  \\
  d = 1 – 22 / (5.83 * 4.47)
  \\
  d = 1–0.844
  \\
  d = 0.156
  $

````python
# тот же пример на python
from scipy.spatial import distance
A = (5, 3)
B = (2, 4)
d = distance.cosine(A, B)
print('Косинусное расстояние:', d)
````

Когда значение косинусного расстояния стремится к нулю, то точки совпадают. Мы в примере выше используем значение: $1 - spatial.distance.cosine(x, y)$, так мы переходим к понятию релевантности: чем значение ближе к единице, тем вероятность выше, что данная строка дата фрейма ближе по смыслу к пользовательскому запросу.

Проверим наше предположение:

````python
strings, relatednesses = strings_ranked_by_relatedness("curling gold medal", df, top_n=5)
for string, relatedness in zip(strings, relatednesses):
    print(f"{relatedness=:.3f}")
    display(string)
````

Как мы видим выше, по запросу золотые медали в керлинге для 5 лучших результатов, мы получили схожести (или релевантности выше 86%). И действительно, в каждом из примеров мы видим содержаться слова `Curling`, `Gold medal game` и множество другой полезной информации.

#### 3. Ask

С помощью функции поиска, описанной выше, мы теперь можем автоматически извлекать необходимые сведения из DataFrame и вставлять их в сообщения для chatGPT.

Ниже мы определяем функцию ask, которая:

* Принимает запрос пользователя
* Ищет в базе знаний текст релевантный запросу
* Вставляет этот текст в сообщение для GPT
* Отправляет сообщение в GPT
* Возвращает ответ GPT

````python
# с этой функцией мы уже знакомы
def num_tokens(text: str, model: str = GPT_MODEL) -> int:
    """Возвращает число токенов в строке для заданной модели"""
    encoding = tiktoken.encoding_for_model(model)
    return len(encoding.encode(text))

# Функция формирования запроса к chatGPT по пользовательскому вопросу и базе знаний
def query_message(
    query: str, # пользовательский запрос
    df: pd.DataFrame, # DataFrame со столбцами text и embedding (база знаний)
    model: str, # модель
    token_budget: int # ограничение на число отсылаемых токенов в модель
) -> str:
    """Возвращает сообщение для GPT с соответствующими исходными текстами, извлеченными из фрейма данных (базы знаний)."""
    strings, relatednesses = strings_ranked_by_relatedness(query, df) # функция ранжирования базы знаний по пользовательскому запросу
    # Шаблон инструкции для chatGPT
    message = 'Use the below articles on the 2022 Winter Olympics to answer the subsequent question. If the answer cannot be found in the articles, write "I could not find an answer."'
    # Шаблон для вопроса
    question = f"\n\nQuestion: {query}"

    # Добавляем к сообщению для chatGPT релевантные строки из базы знаний, пока не выйдем за допустимое число токенов
    for string in strings:
        next_article = f'\n\nWikipedia article section:\n"""\n{string}\n"""'
        if (num_tokens(message + next_article + question, model=model) > token_budget):
            break
        else:
            message += next_article
    return message + question


def ask(
    query: str, # пользовательский запрос
    df: pd.DataFrame = df, # DataFrame со столбцами text и embedding (база знаний)
    model: str = GPT_MODEL, # модель
    token_budget: int = 4096 - 500, # ограничение на число отсылаемых токенов в модель
    print_message: bool = False, # нужно ли выводить сообщение перед отправкой
) -> str:
    """Отвечает на вопрос, используя GPT и базу знаний."""
    # Формируем сообщение к chatGPT (функция выше)
    message = query_message(query, df, model=model, token_budget=token_budget)
    # Если параметр True, то выводим сообщение
    if print_message:
        print(message)
    messages = [
        {"role": "system", "content": "You answer questions about the 2022 Winter Olympics."},
        {"role": "user", "content": message},
    ]
    response = openai.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0 # гиперпараметр степени случайности при генерации текста. Влияет на то, как модель выбирает следующее слово в последовательности.
    )
    response_message = response.choices[0].message.content
    return response_message
````
